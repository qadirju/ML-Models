# Google PageRank Algorithm Implementation

This notebook implements the Google PageRank algorithm on a directed graph using NetworkX. It computes PageRank scores for web pages, visualizes the graph, and exports results to CSV.

## Section 1: Import Required Libraries

Import NetworkX for graph algorithms, Matplotlib for visualization, and CSV for data export. We handle missing dependencies gracefully.

```python
import csv
import sys

try:
    import networkx as nx
except Exception as e:
    print("This script requires networkx. Install with: pip install networkx")
    raise

try:
    import matplotlib.pyplot as plt
except Exception:
    print("matplotlib not installed. Install with: pip install matplotlib")
```

## Section 2: Build the Directed Graph

Create a directed graph using NetworkX with nodes A-F and g1-g5. Add edges representing the hyperlink structure from the assignment diagram (interpreted from the provided figure).

```python
def build_graph():
    """Build a directed graph representing web pages and hyperlinks."""
    G = nx.DiGraph()
    
    # Nodes from the diagram
    nodes = ["A", "B", "C", "D", "E", "F", "g1", "g2", "g3", "g4", "g5"]
    G.add_nodes_from(nodes)
    
    # Edges: these are an interpreted approximation of the provided figure.
    # Update this list to exactly match the assignment image if you have the exact edge set.
    edges = [
        ("A", "B"),
        ("D", "B"),
        ("D", "A"),
        ("E", "B"),
        ("E", "C"),
        ("E", "g1"),
        ("E", "g2"),
        ("g1", "E"),
        ("g2", "E"),
        ("g3", "E"),
        ("g4", "E"),
        ("g5", "E"),
        ("g3", "B"),
        ("g4", "B"),
        ("g5", "B"),
        ("B", "C"),
        ("C", "B"),
        ("F", "E"),
        ("F", "B"),
    ]
    
    G.add_edges_from(edges)
    print(f"Graph created: {len(G.nodes())} nodes, {len(G.edges())} edges")
    return G

# Build the graph
G = build_graph()
```

## Section 3: Compute PageRank Scores

Implement the PageRank algorithm using NetworkX's `pagerank()` function with damping factor (alpha) = 0.85 and convergence tolerance.

```python
def compute_pagerank(G, alpha=0.85, tol=1.0e-06):
    """Compute PageRank scores for all nodes in the graph."""
    pr = nx.pagerank(G, alpha=alpha, tol=tol)
    return pr

# Compute PageRank
pr = compute_pagerank(G)
print("PageRank computation completed.")
```

## Section 4: Display Results

Print PageRank scores for all nodes sorted in descending order, displaying both raw scores and percentage values.

```python
def print_results(pr):
    """Print PageRank results sorted by score in descending order."""
    print("\n" + "="*60)
    print("PageRank Results (sorted by score, descending)")
    print("="*60)
    print(f"{'Node':<8} {'PageRank Score':<20} {'Percentage':<10}")
    print("-"*60)
    for node, score in sorted(pr.items(), key=lambda x: -x[1]):
        print(f"{node:<8} {score:<20.12f} {score*100:>8.2f}%")
    print("="*60)

# Display results
print_results(pr)
```

## Section 5: Visualize the Graph with PageRank

Create a network visualization using Matplotlib and NetworkX where node sizes are proportional to their PageRank scores, with directed edges and labels.

```python
def visualize_pagerank(G, pr, path="pagerank_visualization.png"):
    """Visualize the graph with node sizes proportional to PageRank scores."""
    try:
        import matplotlib.pyplot as plt
    except ImportError:
        print("matplotlib not installed; skipping visualization.")
        return
    
    plt.figure(figsize=(12, 9))
    pos = nx.spring_layout(G, seed=42, k=2, iterations=50)
    
    # Node sizes proportional to PageRank scores
    sizes = [8000 * pr.get(n, 0.001) for n in G.nodes()]
    
    # Draw nodes
    nx.draw_networkx_nodes(G, pos, node_size=sizes, node_color="tab:orange", alpha=0.9)
    
    # Draw edges with arrows
    nx.draw_networkx_edges(G, pos, arrowstyle="->", arrowsize=15, edge_color="gray", width=1.5)
    
    # Draw labels
    nx.draw_networkx_labels(G, pos, font_size=10, font_weight="bold")
    
    plt.title("PageRank Visualization\n(Node size proportional to PageRank score)", fontsize=14)
    plt.axis("off")
    plt.tight_layout()
    plt.savefig(path, dpi=150, bbox_inches="tight")
    print(f"Graph visualization saved to {path}")
    plt.show()

# Visualize the graph
visualize_pagerank(G, pr)
```

## Section 6: Export Results to CSV

Write PageRank results to a CSV file with columns for node name, PageRank score, and percentage, sorted by score in descending order.

```python
def save_results(pr, path="pagerank_results.csv"):
    """Export PageRank results to CSV file."""
    with open(path, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["Node", "PageRank_Score", "Percentage"])
        for node, score in sorted(pr.items(), key=lambda x: -x[1]):
            writer.writerow([node, f"{score:.12f}", f"{score*100:.2f}%"])
    print(f"\nResults exported to {path}")

# Export results
save_results(pr)
```

## Section 7: Summary and Mathematical Background

### Key Concepts:

1. **Random Surfer Model**: PageRank models a web surfer who randomly clicks links with probability `alpha`, or randomly jumps to any page with probability `1-alpha`.

2. **Transition Matrix**: Represents the probability of moving from one page to another. With damping factor alpha=0.85, we follow links 85% of the time and jump randomly 15% of the time.

3. **Eigenvector Problem**: PageRank is the principal eigenvector of the transition matrix, representing the steady-state probability distribution.

4. **Convergence**: The algorithm uses power iteration to compute PageRank until convergence (tolerance = 1e-06).

5. **Interpretation**: Higher PageRank scores indicate pages that are more "important" or more likely to be visited by a random surfer.
